```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.rand(10)

mean = np.mean(data)
median = np.median(data)

print(f"Среднее: {mean:.4f}")
print(f"Медиана: {median:.4f}")
print(f"Разница: {abs(mean - median):.4f}")

plt.figure(figsize=(10, 6))
plt.scatter(range(len(data)), data)

plt.axhline(mean, color='red', linestyle='--')
plt.axhline(median, color='green', linestyle=':')

plt.title('Точечное рассеивание')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import math as m
import matplotlib.pyplot as plt

results = [(m.sqrt(1 + m.exp(m.sqrt(x)) + m.cos(x ** 2))) / abs(1 - m.sin(x) ** 3) + m.log(abs(2 * x)) for x in range(1, 11)]
half = len(results) // 2

fig = plt.figure(figsize=(5, 3))
plt.scatter(range(1, half + 1), results[:half])
plt.grid()
plt.plot(range(1, len(results) + 1), results)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0.0, 11, 1)
y = np.abs(np.cos(x * np.exp(np.cos(x) + np.log(x + 1))))
fig = plt.figure(figsize=(5, 3))
plt.grid()
plt.plot(x, y, c = "r")
plt.fill_between(x, y)
plt.show()
print(np.trapezoid(y))
```

```python
import matplotlib.pyplot as plt
Apple     = [131.96, 121.26, 122.15, 131.46, 124.61, 136.96, 145.86, 151.83, 141.5, 149.8, 165.3, 177.57]
Microsoft = [231.96, 232.38, 235.77, 252.18, 249.68, 270.9, 284.91, 301.88, 281.92, 331.62, 330.59, 336.32]
Google    = [91.37, 101.1, 103.13, 117.68, 117.84, 122.09, 134.73, 144.7, 133.68, 148.05, 141.9, 144.85]
months = ["январь", "февраль", "март", "апрель", "май", "июнь", "июль", "август", "сентябрь", "октябрь", "ноябрь", "декабрь"]
fig = plt.figure(figsize=(7, 4))
plt.plot(months, Apple, label="Apple")
plt.fill_between(months, Apple, alpha=0.1, color='b')
plt.plot(months, Microsoft, label="Microsoft")
plt.fill_between(months, Microsoft, alpha=0.1, color='g')
plt.plot(months, Google, label="Google")
plt.fill_between(months, Google, alpha=0.1, color='r')
plt.grid()
plt.xticks(rotation=34, ha='right')
plt.legend()
plt.title("Динамика стоимости акций за 2021 год")
plt.ylabel("Цена акции (USD)")
plt.show()
```

```python
import math as m
x, y = map(float, input("Введите 2 вещ числа через пробел: ").split())
print("Сложение", x + y)
print("Разность", x - y)
print("Умножение", x * y)
print("Деление", 'на ноль' if not y else x / y)
print("e ^ (x + y) =", m.exp(x + y))
print("sin(x + y) =", m.sin(x + y))
print("cos(x + y) =", m.cos(x + y))
print("x ^ y =", x ** y)
```
